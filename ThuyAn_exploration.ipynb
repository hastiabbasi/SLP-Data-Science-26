```python
import kagglehub

# Download latest version
path = kagglehub.competition_download('spaceship-titanic')

print("Path to competition files:", path)
```

```python
import pandas as pd
import os


train = pd.read_csv(os.path.join(path, "train.csv"))
test = pd.read_csv(os.path.join(path, "test.csv"))
sample_submission = pd.read_csv(
    os.path.join(path, "sample_submission.csv")
)

print(train.shape)
print(test.shape)

display(train.head())
train.info()
```

```python
train["Transported"].value_counts()

```

```python
import matplotlib.pyplot as plt

train["Transported"].value_counts().plot(kind="bar")
plt.title("Number of Transported and Non-Transported Passengers")
plt.xlabel("Transported")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
pd.crosstab(
    train["CryoSleep"],
    train["Transported"],
    normalize="index"
).plot(kind="bar")

plt.title("Transportation Rate by CryoSleep Status")
plt.ylabel("Proportion Transported")
plt.show()
```

```python
pd.crosstab(
    train["Age"],
    train["Transported"],
    normalize="index"
).plot(kind="bar")

plt.title("Transportation Rate by Age Status")
plt.ylabel("Proportion Transported")
plt.show()
```

```python
pd.crosstab(
    train["VIP"],
    train["Transported"],
    normalize="index"
).plot(kind="bar")

plt.title("Transportation Rate by VIP Status")
plt.ylabel("Proportion Transported")
plt.show()
```
